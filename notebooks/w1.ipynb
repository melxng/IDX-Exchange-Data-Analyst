# Week 1 - Monthly Dataset Aggregation

Given monthly MLS files, concatenate all files from January 2024 onwards into combined dataset

```python
import pandas as pd
from pathlib import Path
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
# set path to data folder
data_dir = Path("../data/raw_csv")
```

First, I'll sort the listing csv's by concatenation then filtering based on "Residential" Property Type. Then, I'll do the same thing with the sold csv's and then concatenate the filtered dataframes.

```python
# grab all the "listing" csv files in the data folder
listing_files = sorted(data_dir.glob("CRMLSListing*.csv")) # avoid hard coding the file names
print(len(listing_files), "CRMLS Listing files found:")
listing_files
```

```python
# now grab all the "sold" csv files in the data folder
sold_files = sorted(data_dir.glob("CRMLSSold*.csv"))
print(len(sold_files), "CRMLS Sold files found:")
sold_files
```

### Listing DF
Now we need to read in each file in so we can create a list of the "listing" data to prepare for concatenation.
Additionally, we must confirm that the row counts from before and after concatenation are the same.

```python
# empty list to hold the dataframes for each listing file
listing_df = []

# read in each listing file, then append it to listing_df
for file in listing_files:
    df = pd.read_csv(file, low_memory=False)
    listing_df.append(df)

    # print out name of file and # of rows and cols after each file is read in
    print(file.name, "->", len(df), "ROWS and", len(df.columns), "COLS")
```

```python
# get the total number of rows in all the listing files
listings_rows_before = sum(len(df) for df in listing_df)
print(listings_rows_before)
```

```python
# concatenate all the listing dataframes into one dataframe
listings = pd.concat(listing_df, ignore_index=True)
listings
```

```python
print("# Listing Rows Before Concat:", listings_rows_before)
print("# Listing Rows After Concat:", len(listings))
```

```python
listings["PropertyType"].value_counts()
```

```python
# now filter based on the PropertyType == "Residential"
listings_before_filter = len(listings)

listings_residential = listings[listings["PropertyType"] == "Residential"].copy()

print("# Listing Rows Before 'Residential' Filter: ", listings_before_filter)
print("# Listing Rows After 'Residential' Filter:", len(listings_residential))

listings_residential
```

### Sold DF
Now just do the same thing but with the Sold CSV files. Just follow the same steps.

```python
sold_dfs = []

for file in sold_files:
    df = pd.read_csv(file, low_memory=False)
    sold_dfs.append(df)

    print(file.name, "->", len(df), "ROWS and", len(df.columns), "COLS")
```

```python
# get the total # of rows in all the sold files before concatenation
sold_rows_before = sum(len(df) for df in sold_dfs)
print(sold_rows_before)
```

```python
# concatenate all the sold dataframes into one DF
sold = pd.concat(sold_dfs, ignore_index=True)
sold
```

```python
print("# Sold Rows before Concat:", sold_rows_before)
print("# Sold Rows after Concat:", len(sold))
```

```python
sold["PropertyType"].value_counts()
```

```python
# now filter Sold csv based on the PropertyType == "Residential"
sold_before_filter = len(sold)
sold_residential = sold[sold["PropertyType"] == "Residential"].copy()

print("# Sold Rows Before 'Residential' Filter: ", sold_before_filter)
print("# Sold Rows After 'Residential' Filter:", len(sold_residential))

sold_residential
```

Now that both listings and sold are sorted and concatenated, save both of them as CSVs.

```python
output_dir = Path("../data/processed_csv")
output_dir.mkdir(exist_ok=True)

listings_residential.to_csv(output_dir / "listings_residential.csv", index=False)

sold_residential.to_csv(output_dir / "sold_residential.csv",index=False)
```
